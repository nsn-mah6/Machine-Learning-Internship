```python
# importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# creating a dataframe
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H",
                "I", "J", "K", "L", "M", "N", "O"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 2, 3, 5, 6, 7, 8, 4, 10],
    "Marks": [45, 50, 55, 60, 65, 70, 80, 42, 52, 62, 68, 75, 85, 58, 98],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 55, 68, 78, 82, 88, 95, 72, 100],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male"],
    "Result": ["Fail", "Fail", "Fail", "Pass", "Pass",
               "Pass", "Pass", "Fail", "Fail", "Pass",
               "Pass", "Pass", "Pass", "Pass", "Pass"]
}
df = pd.DataFrame(data)
```

```python
# Visualization: Histogram
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Marks",
    bins=6,
    kde=True
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# visualization: Box Plot
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="Marks"
)

plt.title("Boxplot of Student Marks")
plt.ylabel("Marks")

plt.show()
```

```python
# visualization: Scatter Plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    hue="Result",
    s=100
)

plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# visualization: Count Plot
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Result"
)

plt.title("Student Pass and Fail Distribution")
plt.xlabel("Result")
plt.ylabel("Number of Students")

plt.show()
```

```python
# visualization: Bar Plot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Gender",
    y="Marks",
    errorbar=None
)

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.show()
```

```python
# visualization: Correlation Heatmap
correlation = df[
    ["Study_Hours", "Marks", "Attendance"]
].corr()

print(correlation)
```

```python
# visualization: Heatmap of Correlation Matrix
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Visualization: Pairplot
sns.pairplot(
    df,
    vars=["Study_Hours", "Marks", "Attendance"],
    hue="Result"
)

plt.show()
```
